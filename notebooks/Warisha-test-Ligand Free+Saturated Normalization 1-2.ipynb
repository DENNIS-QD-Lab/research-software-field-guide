```python
import pandas as pd

# Load the CSV file
file_path = 'warisha_array01_o1_Chamber1-2.csv'
df = pd.read_csv(file_path, encoding='UTF-8-SIG')

# Display the head for reference
print(df.head())

# Assuming the Chamber columns are 'Chamber 1' and 'Chamber 2'
# Calculate the global min and max across both chambers
global_min = df[['Chamber 1', 'Chamber 2']].min().min()
global_max = df[['Chamber 1', 'Chamber 2']].max().max()

# Apply min-max normalization to both columns
df['Chamber 1_norm'] = (df['Chamber 1'] - global_min) / (global_max - global_min)
df['Chamber 2_norm'] = (df['Chamber 2'] - global_min) / (global_max - global_min)

# Display head of normalized columns
print(df[['Probe ID', 'Chamber 1_norm', 'Chamber 2_norm']].head())

# Save the normalized dataframe to a new CSV file
normalized_csv = 'normalized_warisha_array01_o1_Chamber1-2.csv'
df.to_csv(normalized_csv, index=False)
print('Normalized CSV saved as:', normalized_csv)
```
