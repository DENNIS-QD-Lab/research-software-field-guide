# Exploring an HDF5 file

This notebook is the interactive twin of `scripts/show_h5_keys.py`. The script is the importable, command-line version; this notebook is where you poke at a file by hand. They share the same `show_keys` function, so the two never drift apart.

To use it: set the path below to one of your `.h5` files, then run all cells (Run > Run All, or the "Restart and Run All" button). Make sure the kernel in the top-right corner says `helper`.

```python
import h5py

from scripts.show_h5_keys import show_keys

# Edit this to point at the file you want to inspect.
h5_file = "path/to/data.h5"
```

## List every group and dataset

`show_keys` walks the whole file and prints its structure. This is the same output you would get by running the script from the terminal.

```python
show_keys(h5_file)
```

## Peek at one dataset

Listing the keys tells you what is in the file. To look at the actual numbers, open the file and index into one dataset by its path. Change `dataset_path` to one of the dataset names printed above.

```python
dataset_path = "ratios/ratio_000/data"

with h5py.File(h5_file, "r") as f:
    data = f[dataset_path][()]  # [()] reads the whole dataset into a NumPy array

print("shape:", data.shape)
print("dtype:", data.dtype)
print("min, max:", data.min(), data.max())
```
